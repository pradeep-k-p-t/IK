```python
# import statements 
import nltk 
nltk.download('pinkt')
nltk.download('punkt')
nltk.download('stopwords')
import pandas as pd 
from sklearn.feature_extraction.text import CountVectorizer
from nltk.tokenize import word_tokenize 
```

```python
corpus = ["This is the first document", "this document is the second document", "and this is the third document", "is this the fourth document"]
print(corpus)
```

```python
# BAG of works 
vectorize = CountVectorizer()
vectors = vectorize.fit_transform(corpus)
print(vectorize.get_feature_names_out())
print(vectors.toarray())
print(type(vectors))
```
